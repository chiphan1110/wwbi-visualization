# Preprocessing

```python
def process_occupation_data(wwbi_data_vietnam):
    # Define the mapping between indicator codes and metadata
    indicator_mapping = {
        # Public sector
        'BI.PWK.PUBS.SN.FE.ZS': {'sector': 'Public', 'occupation': 'Managers', 'level': 'High'},
        'BI.PWK.PUBS.PN.FE.ZS': {'sector': 'Public', 'occupation': 'Professionals', 'level': 'High'},
        'BI.PWK.PUBS.TN.FE.ZS': {'sector': 'Public', 'occupation': 'Technicians', 'level': 'High'},
        'BI.PWK.PUBS.CK.FE.ZS': {'sector': 'Public', 'occupation': 'Clerks', 'level': 'Low'},
        'BI.PWK.PUBS.EO.FE.ZS': {'sector': 'Public', 'occupation': 'Elementary', 'level': 'Low'},
        
        # Private sector
        'BI.PWK.PRVS.SN.FE.ZS': {'sector': 'Private', 'occupation': 'Managers', 'level': 'High'},
        'BI.PWK.PRVS.PN.FE.ZS': {'sector': 'Private', 'occupation': 'Professionals', 'level': 'High'},
        'BI.PWK.PRVS.TN.FE.ZS': {'sector': 'Private', 'occupation': 'Technicians', 'level': 'High'},
        'BI.PWK.PRVS.CK.FE.ZS': {'sector': 'Private', 'occupation': 'Clerks', 'level': 'Low'},
        'BI.PWK.PRVS.EO.FE.ZS': {'sector': 'Private', 'occupation': 'Elementary', 'level': 'Low'}
    }
    
    # Filter data to only include indicators we're interested in
    relevant_indicators = list(indicator_mapping.keys())
    filtered_data = wwbi_data_vietnam[wwbi_data_vietnam['indicator_code'].isin(relevant_indicators)].copy()
    
    # Add metadata columns based on indicator_code
    for column, data in indicator_mapping.items():
        mask = filtered_data['indicator_code'] == column
        for key, value in data.items():
            filtered_data.loc[mask, key] = value
    
    # Define occupation order
    occupation_order = ['Managers', 'Professionals', 'Technicians', 'Clerks', 'Elementary']
    
    # Convert occupation to ordered categorical
    filtered_data['occupation'] = pd.Categorical(
        filtered_data['occupation'],
        categories=occupation_order,
        ordered=True
    )
    
    # Convert sector and level to categorical
    filtered_data['sector'] = pd.Categorical(filtered_data['sector'], categories=['Public', 'Private'])
    filtered_data['level'] = pd.Categorical(filtered_data['level'], categories=['High', 'Low'])
    
    # Convert values to percentages 
    if filtered_data['value'].max() < 1:
        filtered_data['value'] = filtered_data['value'] * 100
    
    return filtered_data
```

# Visualizing

```python
from matplotlib.collections import PolyCollection
import matplotlib.colors as mcolors
import matplotlib.patches as mpatches
import matplotlib.lines as mlines


def gradient_fill(ax, x, y, cmap='gray', alpha=1.0):
    # Normalize y for gradient mapping
    norm = mcolors.Normalize(vmin=min(y), vmax=max(y))
    verts = []
    facecolors = []

    for i in range(len(x) - 1):
        verts.append([
            (x[i], 0), 
            (x[i], y[i]),
            (x[i+1], y[i+1]), 
            (x[i+1], 0)
        ])
        color = plt.get_cmap(cmap)(norm((y[i] + y[i+1]) / 2.0))
        facecolors.append(color)

    poly = PolyCollection(verts, facecolors=facecolors, alpha=alpha, edgecolors='none')
    ax.add_collection(poly)
```

```python
def visualize_level_categories_bar(integrated_df):
    # Calculate representation by level, sector, and year
    level_data = integrated_df.groupby(['sector', 'level', 'year'])['value'].mean().reset_index()
    level_data = level_data[level_data['level'].isin(['High', 'Low'])]
    
    # plt.figure(figsize=(15, 10))
    plt.figure(figsize=(10, 6))

    fig, axes = plt.subplots(2, 1, figsize=(14, 10), sharex=True)
    
    # Color palette 
    # colors = {'High': '#C71000FF' , 'Low': '#008EA0FF'}
    # gap_fill_color = '#666666'  
    colors = {
        'High': '#c2718e',   # Dusty rose/mauve
        'Low': '#75b79e'     # Sage teal
    }
    gradient_cmap = 'YlOrBr'  # Yellow-Orange-Brown gradient

    
    # Plot for each sector
    for i, sector in enumerate(['Public', 'Private']):
        sector_data = level_data[level_data['sector'] == sector]
        
        # Get all years 
        years = sorted(sector_data['year'].unique())
        
        # Calculate bar positions
        bar_width = 0.35
        x_positions = np.arange(len(years))
        
        # Plot High and Low level bars
        for j, level in enumerate(['High', 'Low']):
            level_subset = sector_data[sector_data['level'] == level]
            
            # Create a mapping from year to value
            year_to_value = {row['year']: row['value'] for _, row in level_subset.iterrows()}
            
            # Get values for all years (use 0 if data is missing)
            values = [year_to_value.get(year, 0) for year in years]
            
            # Plot the bars
            bar_positions = x_positions + (j - 0.5) * bar_width
            bars = axes[i].bar(bar_positions, values, width=bar_width, 
                      label=f'{level}-Level', color=colors[level], alpha=0.85)
            for bar in bars:
                
                # bar styling
                bar.set_linewidth(0)
                bar.set_edgecolor('none')
                bar.set_zorder(3)
                

                # Add percentage labels on top of the bars
                height = bar.get_height()
                axes[i].annotate(f'{height:.1f}%',
                                xy=(bar.get_x() + bar.get_width() / 2, height),
                                xytext=(0, 3), textcoords="offset points",
                                ha='center', va='bottom', fontsize=10, color='#333', fontweight='bold')
                        
        
        # Calculate and plot the gap between high and low
        high_values = sector_data[sector_data['level'] == 'High'].set_index('year')['value']
        low_values = sector_data[sector_data['level'] == 'Low'].set_index('year')['value']
        
        # Create a complete index of all years
        all_years = pd.Index(years)
        high_values = high_values.reindex(all_years).fillna(0)
        low_values = low_values.reindex(all_years).fillna(0)
        
        # Calculate the gap
        gap_values = high_values - low_values
        
        # Fill area under gap line
        gradient_fill(axes[i], x_positions, gap_values.values, cmap=gradient_cmap, alpha=0.3)
        
        # Plot the gap as a line
        axes[i].plot(x_positions, gap_values, 'k--', marker='o', color='Orange', label='Gap (High - Low)')
        
        # Subfigure Title
        axes[i].set_title(f'{sector} Sector', 
                          fontweight='bold', 
                          fontsize=16,
                          pad=10,
                          color='#333333')
        
        # Label ------------------------------------------------
        axes[i].set_ylabel('Female Representation (%)', fontsize=14)
        axes[i].set_xticks(x_positions)
        axes[i].tick_params(axis='x', labelsize=14, labelbottom=True)
        axes[i].set_xticklabels(years)
        
        # Add a horizontal line at 0 for the gap reference
        axes[i].axhline(y=0, color='gray', linestyle='-', alpha=0.3)
        
        # Grid ---------------------------------------------------------------------
        # Grid styling
        # axes[i].grid(axis='y', linestyle='--', alpha=0.7)
        axes[i].yaxis.grid(True, linestyle='--', linewidth=0.5, color='#E0E0E0', alpha=0.8)
        axes[i].grid(True, which='major', linewidth=1., color='#E0E0E0', alpha=1)
        axes[i].grid(True, which='minor', linewidth=0.5, color='#EEEEEE', alpha=0.8, linestyle='--')
        
        # Ensure grid lines are behind the data
        axes[i].set_axisbelow(True)
    
    # Legend ------------------------------------------------
    
    high_patch = mpatches.Patch(color=colors['High'], label='High-Level')
    low_patch = mpatches.Patch(color=colors['Low'], label='Low-Level')
    gap_line = mlines.Line2D([], [], color='Orange', linestyle='--', marker='o', label='Gap (High - Low)')

    legend = fig.legend(handles=[high_patch, low_patch, gap_line],
                        loc='lower center', ncol=3,
                        frameon=True, 
                        fontsize=12, 
                        bbox_to_anchor=(0.5, 0))
    
    legend.get_title().set_fontweight('bold')
    
    # Remove individual legends
    for ax in axes:
        if ax.get_legend():
            ax.get_legend().remove()
    
    # Adjust layout
    plt.tight_layout(rect=[0, 0.05, 1, 0.95])
    plt.subplots_adjust(top=0.90, hspace=0.3, wspace=0.25)
    plt.suptitle('Female Representation by Occupation Level in Vietnam (2010-2016)', 
                 fontsize=20, 
                 fontweight='bold')
    plt.savefig('../viz/female_occupation_gap_combined.png', dpi=600, bbox_inches='tight')

    
    plt.show()
```

```python
from matplotlib.lines import Line2D
def visualize_all_categories_lollipop(integrated_df):
    # Get average value by sector, occupation, and year
    grouped = integrated_df.groupby(['sector', 'occupation', 'year'])['value'].mean().reset_index()

    plt.figure(figsize=(10, 6))
    fig, axes = plt.subplots(2, 1, figsize=(15, 12), sharex=True)

    # Color palette for occupations
    occupations = grouped['occupation'].astype('category').cat.categories
    color_scheme = ['#4a6da7', '#7cb07c', '#b5825b', '#a279a7', '#d87d7d']    
    occupation_colors = {occupation: color for occupation, color in zip(occupations, color_scheme)}
    colors = [occupation_colors[occ] for occ in occupations]

    for i, sector in enumerate(['Public', 'Private']):
        sector_data = grouped[grouped['sector'] == sector]
        years = sorted(sector_data['year'].unique())

        # Adjust spacing between occupations
        x_positions = np.arange(len(years))
        offset_width = 0.12  

        for j, occupation in enumerate(occupations):
            occ_data = sector_data[sector_data['occupation'] == occupation]
            year_to_value = {row['year']: row['value'] for _, row in occ_data.iterrows()}
            values = [year_to_value.get(year, 0) for year in years]
            pos = x_positions + (j - len(occupations)/2 + 0.5) * offset_width

            # Draw vertical line (stick)
            axes[i].vlines(x=pos, ymin=0, ymax=values, 
                           color=colors[j], 
                           alpha=1, 
                           linewidth=8)

            # Draw the circle (head)
            axes[i].scatter(pos, values,
                            color='white',
                            s=900,
                            alpha=0.3,
                            edgecolors=colors[j],
                            zorder=2)
            axes[i].scatter(pos, values,
                            color=colors[j],
                            s=600,
                            edgecolors='white',
                            linewidth=1,
                            alpha=0.9,
                            label=occupation,
                            zorder=3)

            # # Add value label
            for (x, y) in zip(pos, values):
                    axes[i].text(x, y, f'{int(round(y))}',
                    ha='center', va='center',
                    color='white', fontsize=10, fontweight='bold', zorder=4)
                

        # Title and formatting
        axes[i].set_title(f'{sector} Sector', fontsize=16, fontweight='bold')
        axes[i].set_ylabel('Female Representation (%)')
        axes[i].set_xticks(x_positions)
        axes[i].tick_params(axis='x', labelsize=12, labelbottom=True)
        axes[i].set_xticklabels(years, fontsize=12)
        
        # Set y-axis to show percentages correctly
        y_max = int(grouped['value'].max() // 10 + 2) * 10
        axes[i].set_ylim(0, y_max)
        axes[i].grid(axis='y', linestyle='--', alpha=0.7)
        
                
        # Grid styling
        axes[i].set_axisbelow(True)
        axes[i].grid(which='major', axis='y', linestyle='--', linewidth=1, color='#E0E0E0', alpha=0.7)
        axes[i].grid(which='both', axis='x', linestyle='', linewidth=0)

    # Legend
    legend_handles = [
        Line2D([0], [0], marker='o', color='w', label=occ, markerfacecolor=occupation_colors[occ], 
            markersize=10, markeredgecolor=occupation_colors[occ])
        for occ in occupations
    ]

    legend = fig.legend(legend_handles, occupations, 
                        fontsize=12, title_fontsize=14,
                        loc='lower center', ncol=5,
                        frameon=True,  
                        bbox_to_anchor=(0.45, 0))
    
    legend.get_title().set_fontweight('bold')

    plt.tight_layout(rect=[0, 0.05, 0.85, 0.95])
    plt.subplots_adjust(top=0.90, hspace=0.3)
    plt.suptitle('Female Representation by Occupation Type in Vietnam (2010-2016)', 
                 fontsize=20, fontweight='bold', x=0.45)
    
    plt.savefig('../viz/female_occupation_lollipop.png', dpi=600, bbox_inches='tight')

    plt.show()
```

```python
visualize_all_categories_lollipop(occupation_data)
```
